# MediXAI — Dataset Exploration

Reproducible EDA for the Healthcare Symptoms–Disease Classification dataset.

> Run from the `ml/` directory. Place the real Kaggle CSV in `ml/data/` (or set `$MEDIXAI_DATASET`). If no real dataset is present, run `python -m src.synthetic` first — the generated demo data is clearly labelled SYNTHETIC.

```python
import sys, os
sys.path.insert(0, os.getcwd())
from src.data_loader import load_dataset, catalog_dataset
```

```python
df, cat, used = load_dataset()
print('Source      :', used)
print('Shape       :', df.shape)
print('Columns     :', list(df.columns))
print('Label       :', cat.label)
print('Symptom mode:', cat.symptom_mode)
print('Dtypes      :\n', df.dtypes)
print('Missing     :\n', df.isna().sum())
print('Duplicates  :', df.duplicated().sum())
print('Classes     :', df[cat.label].nunique(), 'diseases')
df.head()
```

```python
print(df[cat.label].value_counts().to_string())
```

```python
import pandas as pd
from src.preprocessing import build_symptom_text
symptom_text = build_symptom_text(df, cat)
freq = pd.Series([t for row in symptom_text.str.split(' ') for t in row if t]).value_counts()
print('Unique symptom tokens:', len(freq))
freq.head(25)
```

```python
counts = symptom_text.str.split(' ').apply(len)
print('Symptom count  min/mean/max:', counts.min(), round(counts.mean(),2), counts.max())
counts.hist(bins=range(0, counts.max()+2), align='left')
```
